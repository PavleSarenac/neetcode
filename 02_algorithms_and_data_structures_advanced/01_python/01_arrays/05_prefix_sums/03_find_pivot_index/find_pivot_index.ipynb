```python
class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        prefixSums = [nums[0]]
        for i in range(1, len(nums)):
            prefixSums.append(prefixSums[i - 1] + nums[i])

        suffixSums = [nums[len(nums) - 1]]
        for i in range(len(nums) - 2, -1, -1):
            suffixSums.insert(0, suffixSums[0] + nums[i])

        for i in range(len(nums)):
            leftSum = rightSum = 0
            if i > 0:
                leftSum = prefixSums[i - 1]
            if i < len(nums) - 1:
                rightSum = suffixSums[i + 1]
            if leftSum == rightSum:
                return i
        return -1
```

```python
class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        n = len(nums)
        for i in range(n):
            leftSum = rightSum = 0
            for l in range(i):
                leftSum += nums[l]
            for r in range(i + 1, n):
                rightSum += nums[r]
            if leftSum == rightSum:
                return i
        return -1
```

```python
class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        n = len(nums)
        prefixSum = [0] * (n + 1)
        for i in range(n):
            prefixSum[i + 1] = prefixSum[i] + nums[i]

        for i in range(n):
            leftSum = prefixSum[i]
            rightSum = prefixSum[n] - prefixSum[i + 1]
            if leftSum == rightSum:
                return i
        return -1
```

```python
class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        total = sum(nums)
        leftSum = 0
        for i in range(len(nums)):
            rightSum = total - nums[i] - leftSum
            if leftSum == rightSum:
                return i
            leftSum += nums[i]
        return -1
```
