```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        writeIndex = 0
        for readIndex in range(len(nums)):
            if writeIndex > 0 and nums[writeIndex - 1] == nums[readIndex]:
                continue
            nums[writeIndex] = nums[readIndex]
            writeIndex += 1
        return writeIndex
```

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        n = len(nums)
        l = r = 0
        while r < n:
            nums[l] = nums[r]
            while r < n and nums[r] == nums[l]:
                r += 1
            l += 1
        return l
```
